## Ejercicio 12
¿Existe alguna relación lineal entre las variables numéricas del dataset? Intenta pintar este razonamiento mediante un gráfico

```python
import seaborn as sns
import matplotlib.pyplot as plt


# Calcular correlaciones
corr = beer1.corr(numeric_only= True)

# Visualizar
plt.figure(figsize=(8, 6))
sns.heatmap(corr, 
            annot=True, 
            fmt='.2f',           # Formato 2 decimales
            cmap='RdBu_r',       # Rojo-Azul invertido
            center=0,
            square=True,
            linewidths=1,
            cbar_kws={'label': 'Correlación'})

plt.title('Matriz de Correlación - Variables Numéricas', fontsize=16)
plt.tight_layout()
plt.show()
```

## Ejercicio 13
Representa el nivel de alcohol (columna `abv`) en un histograma. ¿Se trata de una distribución simétrica o asimétrica?

```python
plt.figure(figsize=(10, 6))
plt.hist(beer1['abv'], bins=20, edgecolor='black')
plt.xlabel('Nivel de Alcohol (ABV)')
plt.ylabel('Frecuencia')
plt.title('Distribución del Nivel de Alcohol')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
# Calcular estadísticas
media = beer1['abv'].mean()
mediana = beer1['abv'].median()
skewness = beer1['abv'].skew()

print(f"Media: {media:.3f}")
print(f"Mediana: {mediana:.3f}")
print(f"Asimetría (skewness): {skewness:.3f}")

# Interpretación del skewness:
# skewness ≈ 0: distribución simétrica
# skewness > 0: asimétrica positiva (cola a la derecha)
# skewness < 0: asimétrica negativa (cola a la izquierda)
```
