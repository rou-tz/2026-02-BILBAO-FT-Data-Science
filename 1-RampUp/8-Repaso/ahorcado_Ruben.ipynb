## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python
#Comenzamos creando la funcion ahorcado, y declaramos errores: ira almacenando el numero de errores, palabra: tendra la palabra a adivinar 
# y por ultimo letras_ok: esta variable ira almacenando las letras correctas de la palabra.

def ahorcado():
    
    errores = 0
    palabra = "BOOTCAMP"
    letras_ok = []
    
    print(f"Vamos a jugar al ahorcado, mi palabra tiene 8 letras. Suerte!!")

#con un while de errores menor que 6 hacemos que el bucle funcione hasta que el numero de errores sea 6 y pare.
    while errores < 6:
        '''
        Para hacer el casillero creamos una lista que he llamado resultado. Entonces por cada letra que introduzca el usuario, recorremos la palabra y si esta 
        incluida en la lista letras_ok, añadimos esa letra a la lista resultado, y si no esta mantenemos el "_"
        ''' 
        resultado = []
        for j in palabra:
            if j in letras_ok:
                resultado.append(j)
            else:
                resultado.append("_")
    
        #Esta es la manera de hacer lo mismo que arriba pero en comprehension // resultado = " ".join(j if j in letras_ok else "_" for j in palabra)
        print("Palabra :", *resultado)
        print("Errores :", errores,"/6")
        
        #Este será el mensaje que saldra si acertamos la palabra completa.
        if not "_" in resultado:
            print(f"BIEEEN! Has ganado!!, La palabra era: {palabra}")
            return
       
        #Pedimos la letra al usuario
        letra = input("Introduce una letra").upper()
          
        ''' Hacemos la comparacion de letras. Tenemos la lista letras_ok que ira almacenando que letras son correctas. Si pones una letra correcta que ya se haya 
        utilizado devolvera un mensaje, sino añadimos esa letra a la lista letras_ok. Tambien en caso de fallo sumaremos 1 a la variable errores.
        '''
        if letra in letras_ok:
            print(f"Ya has usado esta letra")
        elif letra in palabra:
            letras_ok.append(letra)
            print(f"¡Bien! la letra {letra} está en la palabra.")
        else:
            errores += 1
            print(f"'{letra}' no está. Errores: {errores}/6")

    print(f"\nFin del juego. La palabra era: {palabra}")


ahorcado()
        
````
