## Ejercicio Extra
Programa una calculadora que

* Pregunte qué operación quieres realizar (multiplicación/división/suma/resta/potencia/salir)
* Pida dos números 
* Realice la operación
* Siga funcionando hasta que el usuario elija salir

#### Bonus:
* Que guarde todas las operaciones realizadas y las imprima por pantalla cuando el usuario elija esa opción
* Que sea robusta frente a entradas erróneas del usuario o frente a operaciones como dividir por 0

```python
import math
import pprint
import time
```

```python
resultados = []
activa = True

#No es un programa a prueba de entradas erroneas.
#Queremos que funcione mientras activa este en True, y le pedimos al usuario que intruduzca que operacion quiere realizar. Debera escribir el nombre de la operacion en el cuadro
while activa:
    operacion = input("Hola, introduzca la operacion que desea realizar:  multiplicar, dividir, sumar, restar, potencia, resultados o salir")
    #Una de las opciones es salir, para parar la maquina
    if operacion == "salir":
            print("Muchas gracias")
            break
    #Si escribe la operacion resultados le mostramos la lista de resultados que hemos ido almacenando        
    elif operacion == "resultados":
        print("Estos son todos los resultados", resultados)
    
    #Aqui comenzamos con las operaciones, si escribe algo diferente de "salir" le pedimos dos numeros para continuar    
    elif operacion != "salir":
        num_1 = float(input("Introduzca el primer numero").replace(",","."))
        num_2 = float(input("Introduzca el segundo valor").replace(",","."))
       #De aqui en adelante vienen las operaciones y el almacenamiento de los resultados. 
        if operacion == "multiplicar":
            multiplicacion = num_1 * num_2
            resultados.append(multiplicacion)
            print("El resultado de la multiplicacion", multiplicacion)
            
        elif operacion == "dividir":
            division = num_1 / num_2
            resultados.append(division)
            print("El resultado de la division es", division)
            
        elif operacion == "sumar":
            suma = num_1 + num_2
            resultados.append(suma)
            print("El resultado de la suma es", suma)
            
        elif operacion == "restar":
            resta = num_1 - num_2
            resultados.append(resta)
            print("El resultado de la resta es", resta)
            
        elif operacion == "potencia":
            potencia = num_1**num_2
            resultados.append(potencia)
            print("El resultado de la potencia es", potencia)    
```
