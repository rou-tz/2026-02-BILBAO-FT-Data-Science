```python
#esta funcion sera pedida en disparar(tablero) y pide al usuario un valor entre el 0 y 9 para las coordenadas.
def coordenada(nombre):
    while True:
        valor = int(input(f"Introduce {nombre} (0-9): "))
        if 0 <= valor <= 9:
            return int(valor)
        else:
            print(f"El '{valor}' no es válido. Escribe un número entre 0 y 9.")


#Esta es la funcion de disparar que pide las coordenadas la funcion coordenada, y segun lo que introduzca cambia las posiciones por "x" de tocado, o por "o" de agua
def disparar(tablero):
    
    x = coordenada("la fila   ")
    y = coordenada("la columna")
        
    if tablero[x][y] == "B":
        tablero[x][y] = "x"
        print(f"\nTocado!, Posicion: {x},{y}")
        return True
    
    elif tablero [x][y] == "~":
        tablero[x][y] = "o"
        print(f"\nOhh lo siento, Agua")
        return False
    elif tablero [x][y] in ("x","o"):
        print(f"Ya has introducido esta coordenada")
        return False

#con este bucle iniciamos el juego, llamamos a disparar, y mientras siga devolviendo True nos imprime el turno extra y el tablero.

while disparar(tablero):
    print("¡Turno extra! Vuelves a disparar.\n")
    print()
    imprimir_tablero(tablero)

#este print es para el momento en que salimos del bucle por haber fallado el tiro.
print()
imprimir_tablero(tablero)




```
