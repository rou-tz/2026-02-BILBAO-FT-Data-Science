### DATASETS Y FUENTES ALTERNATIVAS DE DATOS

**Dataset utilizado:** IBM Telco Customer Churn

- Servicios contratados (telefonía, internet, seguridad online, soporte técnico, streaming...)
- Información del contrato (tipo de contrato, método de pago, factura electrónica)
- Datos de facturación (cargo mensual, cargo total acumulado, meses de permanencia)
- **Variable target:** `Churn` — si el cliente abandonó el servicio el último mes
